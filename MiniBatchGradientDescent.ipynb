```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
reg=LinearRegression()
reg.fit(X_train,y_train)
y_pred=reg.predict(X_test)
print(r2_score(y_test,y_pred))
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
class MBGDRegressor:

  def __init__(self,batch_size,learning_rate=0.01,epochs=100):
    self.batch_size=batch_size
    self.lr=learning_rate
    self.epochs=epochs
    self.coef_=None
    self.intercept_=None

  def fit(self,X_train,y_train):
    self.intercept_=0
    self.coef_=np.ones(X_train.shape[1])

    for i in range(self.epochs):
      for j in range(int(X_train.shape[0]/self.batch_size)):#will work till batch size
        idx=np.random.randint(low=0,high=X_train.shape[0],size=self.batch_size)
        y_hat=np.dot(X_train[idx],self.coef_)+self.intercept_
        der_intercept=-2*np.mean(y_train[idx]-y_hat)
        der_coef=-2*np.dot((y_train[idx]-y_hat),X_train[idx])
        self.intercept_=self.intercept_-(self.lr*der_intercept)
        self.coef_=self.coef_-(self.lr*der_coef)
    print(self.intercept_,self.coef_)

  def predict(self,X_test):
    return np.dot(X_test,self.coef_)+self.intercept_

```

```python
mbgd=MBGDRegressor(batch_size=int(X_train.shape[0]/50),learning_rate=0.01,epochs=100)
mbgd.fit(X_train,y_train)
mbgd.predict(X_test)
print(r2_score(y_test,mbgd.predict(X_test)))
```

```python
#We dont have MBGDR inbuilt , we use SGDRegressor like this
from sklearn.linear_model import SGDRegressor
sgd=SGDRegressor(learning_rate='constant',eta0=0.1)
batch_size=35

for i in range(100):
  idx=np.random.randint(low=0,high=X_train.shape[0],size=batch_size)
  sgd.partial_fit(X_train[idx],y_train[idx])

y_pred=sgd.predict(X_test)
print(r2_score(y_test,y_pred))

```
