```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=2)
```

```python
plt.scatter(X,y)
```

```python
lr=LinearRegression()
lr.fit(X,y)
```

```python
print(lr.coef_) #m value
```

```python
print(lr.intercept_) #c value
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')#The actual line from OLS,gradient descent line should be close to this
```

```python
#Lets find Gradient Descent now
#suppose m is 93.15
#Then we chose a random b ie 0 for now
#y=mx+b

m=93.15
y_pred=m*X+0
plt.scatter(X,y)
plt.plot(X,y_pred,color='blue',label='b=0')
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.legend()
plt.show()
#blue line should be as close to red as possible, because RED is the OLS solution
```

```python
b=0
m=93.15
loss_slope=-2 *np.sum(y-m*X.ravel()-b)
print(loss_slope)  #slope value
```

```python
#Find step size ie Learning rate*slope
learning_rate=0.1
step_size=loss_slope*learning_rate
print(step_size)
```

```python
#Now finding new b
b=b-step_size
print(b)
#value is negative so we should increment
```

```python
#But drawing the fig again
m=93.15
y_pred=m*X+0
y_pred1=m*X+b
plt.scatter(X,y)
plt.plot(X,y_pred,color='blue',label='b=0')
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='green',label='b='+str(b))
plt.legend()
plt.show()
```

```python
#Now finding new b
b=b-step_size
print(b)
```
