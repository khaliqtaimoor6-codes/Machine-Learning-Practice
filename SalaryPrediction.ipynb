```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

df=pd.read_csv("https://raw.githubusercontent.com/AnnaShestova/salary-years-simple-linear-regression/master/Salary_Data.csv")
X=df[['YearsExperience']] # Make X a DataFrame (2D)
y=df['Salary']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2) # Correct assignment order

lr=LinearRegression()

# print(X_train.shape)
# print(y_train.shape)
# print(X_test.shape)
# print(y_test.shape)

lr.fit(X_train,y_train)

print(X_test)
print(y_test)
# Use X_test directly as it's already 2D (DataFrame)
print(lr.predict(X_test.iloc[[0]])) # Pass a DataFrame with a single row for prediction

print(lr.predict(X_test))
print(lr.score(X_test,y_test))
```

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score

# We need to give only 2 factors: Y prediction , Y test values
# Y is obviously the output column and test means for tesing it was separated

y_pred=lr.predict(X_test)
print(y_pred)

print("MAE :",mean_absolute_error(y_test,y_pred)) #MAE has same units
print("MSE :", mean_squared_error(y_test,y_pred))
print("RMSE :", np.sqrt(mean_squared_error(y_test,y_pred)))
print("R2 Score :", r2_score(y_test,y_pred))
print("Adjusted R2 Score :", 1-(1-r2_score(y_test,y_pred))*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1))


```
