```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
X,y=make_regression(n_samples=100,n_features=1,n_informative=1, n_targets=1, noise=20,random_state=2)
```

```python
plt.scatter(X,y)
```

```python
ler=LinearRegression()
X_train, X_test,y_train, y_test=train_test_split(X,y,test_size=0.2,random_state=2)
ler.fit(X_train,y_train)
print(ler.coef_)
print(ler.intercept_)
y_pred=ler.predict(X_test)
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)


```

```python
class GDRegressor:

  def __init__(self,learning_rate,epochs):
    self.m=100
    self.b=-120
    self.lr=learning_rate
    self.epochs=epochs

  def fit(self,X,y):
    #Calculate b using GD
    for i in range(self.epochs):
      slope_loss_b=-2*np.sum(y-self.m*X.ravel()-self.b)
      self.b=self.b-(self.lr*slope_loss_b)
      slope_loss_m=-2*np.sum(X.ravel()*(y-self.m*X.ravel()-self.b))

      self.m=self.m-(self.lr*slope_loss_m)
    print(self.b , self.m)

  def predict(self,X):
    return self.m*X+self.b

```

```python
gd=GDRegressor(0.001,101)
#if you see the valye not being reached ie b here
#decrease the learning rate and increase the epochs
gd.fit(X_train,y_train)


y_pred=gd.predict(X_test)
from sklearn.metrics import r2_score
print(f"R2 score : { r2_score(y_test,y_pred)}")
```
